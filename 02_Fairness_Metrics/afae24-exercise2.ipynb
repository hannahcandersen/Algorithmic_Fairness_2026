## Task 3 (Fairness/Utility-Tradeoff)

**Remark:** You might find it easier to solve this task by writing some code.

Look at the following two picture that presents group and target as usual, but instead of seeing the outcome of the selection, we just see the score. (You can think about it as the grade point average achieved in school.)

![](002.png)

We discuss a classifier that selects all individuals above a certain (maybe group-specific) threshold.

1) Sketch the ROC curve for a threshold based classifier (or actually plot it).
2) Let us say that we are in a situation where a false positive costs us 150 DKK, a true positive gives us 100 DKK.
    - If we want to maximize the profit using a single threshold, which one would it be?
        - At 9,7 profit is 200
    - If we want to achieve statistical parity by choosing individual threshold values, which thresholds can we choose? Which one provides the best utility, i.e., the largest profit? 
    - At which individual thresholds do you achieve equalized odds in this setting? Which setting achieves the best utility?

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# 1) Data from the picture
# -----------------------------
# G=0: triangles, G=1: circles
# T=1: blue, T=0: red
data = [
    # Triangles (G=0)
    {"G": 0, "T": 1, "score": 9.7},
    {"G": 0, "T": 1, "score": 7.6},
    {"G": 0, "T": 1, "score": 10.1},
    {"G": 0, "T": 0, "score": 9.0},
    {"G": 0, "T": 0, "score": 9.5},
    {"G": 0, "T": 0, "score": 9.1},

    # Circles (G=1)
    {"G": 1, "T": 1, "score": 6.8},
    {"G": 1, "T": 1, "score": 6.7},
    {"G": 1, "T": 1, "score": 8.8},
    {"G": 1, "T": 0, "score": 6.6},
    {"G": 1, "T": 0, "score": 6.4},
    {"G": 1, "T": 0, "score": 7.5},
]
df = pd.DataFrame(data)

# -----------------------------
# 2) Helpers: thresholds, metrics, ROC
# -----------------------------
def candidate_thresholds(scores):
    """
    For a threshold rule score >= tau:
    the classification only changes when tau passes a score value.
    So we test midpoints between sorted unique scores, plus +/-inf.
    """
    uniq = np.sort(np.unique(scores))
    taus = [np.inf]
    taus += [(uniq[i] + uniq[i+1]) / 2 for i in range(len(uniq) - 1)]
    taus += [-np.inf]
    return taus

def confusion_metrics(subdf, tau):
    """
    Compute TP, FP, TN, FN, TPR, FPR, selected count, and profit for threshold tau.
    """
    sel = subdf["score"] >= tau  # selection rule
    TP = int(((sel) & (subdf["T"] == 1)).sum())
    FP = int(((sel) & (subdf["T"] == 0)).sum())
    FN = int(((~sel) & (subdf["T"] == 1)).sum())
    TN = int(((~sel) & (subdf["T"] == 0)).sum())

    TPR = TP / (TP + FN) if (TP + FN) else 0.0
    FPR = FP / (FP + TN) if (FP + TN) else 0.0

    # Utility / profit model:
    # +100 per true positive, -150 per false positive
    profit = 100 * TP - 150 * FP

    return {
        "tau": tau,
        "TP": TP, "FP": FP, "TN": TN, "FN": FN,
        "TPR": TPR, "FPR": FPR,
        "selected": int(sel.sum()),
        "profit": profit,
    }

def roc_points(subdf):
    taus = candidate_thresholds(subdf["score"].values)
    pts = [confusion_metrics(subdf, tau) for tau in taus]
    # Sort by FPR to make a clean ROC curve
    pts = sorted(pts, key=lambda d: (d["FPR"], d["TPR"]))
    return pts

# -----------------------------
# 3) Plot ROC (overall + per group)
# -----------------------------
overall_pts = roc_points(df)
g0_pts = roc_points(df[df["G"] == 0])
g1_pts = roc_points(df[df["G"] == 1])

plt.figure()
plt.plot([p["FPR"] for p in overall_pts], [p["TPR"] for p in overall_pts], marker="o", label="Overall")
plt.plot([p["FPR"] for p in g0_pts],      [p["TPR"] for p in g0_pts],      marker="o", label="G=0 triangles")
plt.plot([p["FPR"] for p in g1_pts],      [p["TPR"] for p in g1_pts],      marker="o", label="G=1 circles")
plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.title("ROC curve for threshold classifier (score >= τ)")
plt.legend()
plt.grid(True)
plt.show()

# -----------------------------
# 4) Best single threshold for profit
# -----------------------------
taus_all = candidate_thresholds(df["score"].values)
best_single = max((confusion_metrics(df, tau) for tau in taus_all), key=lambda d: d["profit"])

print("\nBest SINGLE threshold (overall):")
print(best_single)

# -----------------------------
# 5) Group-specific thresholds for Statistical Parity
#    Statistical parity here means same selection rate across groups.
#    Since each group has 6 people, this is equivalent to selecting the SAME COUNT in each group.
# -----------------------------
taus_g0 = candidate_thresholds(df[df["G"] == 0]["score"].values)
taus_g1 = candidate_thresholds(df[df["G"] == 1]["score"].values)

parity_candidates = []
for t0 in taus_g0:
    m0 = confusion_metrics(df[df["G"] == 0], t0)
    for t1 in taus_g1:
        m1 = confusion_metrics(df[df["G"] == 1], t1)
        if m0["selected"] == m1["selected"]:
            parity_candidates.append({
                "tau_g0": t0, "tau_g1": t1,
                "selected_each": m0["selected"],
                "profit_total": m0["profit"] + m1["profit"],
                "m0": m0, "m1": m1,
            })

best_parity = max(parity_candidates, key=lambda d: d["profit_total"])

print("\nBest group-specific thresholds satisfying STATISTICAL PARITY:")
print({
    "tau_g0": best_parity["tau_g0"],
    "tau_g1": best_parity["tau_g1"],
    "selected_each": best_parity["selected_each"],
    "profit_total": best_parity["profit_total"],
})
print("Details G=0:", best_parity["m0"])
print("Details G=1:", best_parity["m1"])

# -----------------------------
# 6) Group-specific thresholds for Equalized Odds
#    Equalized odds: same TPR AND same FPR across groups.
# -----------------------------
eo_candidates = []
for t0 in taus_g0:
    m0 = confusion_metrics(df[df["G"] == 0], t0)
    for t1 in taus_g1:
        m1 = confusion_metrics(df[df["G"] == 1], t1)
        if np.isclose(m0["TPR"], m1["TPR"]) and np.isclose(m0["FPR"], m1["FPR"]):
            eo_candidates.append({
                "tau_g0": t0, "tau_g1": t1,
                "TPR": m0["TPR"], "FPR": m0["FPR"],
                "profit_total": m0["profit"] + m1["profit"],
                "m0": m0, "m1": m1,
            })

best_eo = max(eo_candidates, key=lambda d: d["profit_total"])

print("\nBest group-specific thresholds satisfying EQUALIZED ODDS:")
print({
    "tau_g0": best_eo["tau_g0"],
    "tau_g1": best_eo["tau_g1"],
    "TPR": best_eo["TPR"],
    "FPR": best_eo["FPR"],
    "profit_total": best_eo["profit_total"],
})
print("Details G=0:", best_eo["m0"])
print("Details G=1:", best_eo["m1"])
```

```python
from folktables.acs import adult_filter
from folktables import ACSDataSource, BasicProblem
import numpy as np
from sklearn.model_selection import train_test_split


data_source = ACSDataSource(survey_year='2018', horizon='1-Year', survey='person')
acs_data = data_source.get_data(states=["CA"], download=True)

ACSIncomeNew = BasicProblem(
    features=[
        'AGEP', # include AGE
        'COW', # include class of worker
        'SCHL', # include school education
        'WKHP', # include reported working hours
        'SEX', # include sex
    ],
    target='PINCP', # personal income 
    target_transform=lambda x: x > 25000, # T=1 if income > 25k, else T=0
    group='SEX', # G=1 if male, G=2 if female
    preprocess=adult_filter, # removes minors and invalid data
    postprocess=lambda x: np.nan_to_num(x, -1), # replace any remaining NaNs with -1 (if any)
)

features, label, group = ACSIncomeNew.df_to_numpy(acs_data) # convert to numpy arrays

X_train, X_test, y_train, y_test, group_train, group_test = train_test_split( # split into train and test sets
    features, label, group, test_size=0.2, random_state=0)
```

## Task 3 (Building and evaluation a classifier)

Use your knowledge from Task 2 to build a simple classifier to predict whether an individual in the test group will make more than 25000 USD or not per year. Try to keep your classifier simple, and see whether you can include some kind of threshold.

1) Evaluate the accuracy of your classifier(s).
2) Check the following fairness conditions of your classifier for your classifier:
   - Statistical Parity ($G \perp S$)
   - Equalized odds ($G \perp S | T)$, report both $T = 1$ (true positive rate) and $T = 0$ (false positive rate).
   - Equalized outcome ($G \perp T | S)$, report both $S = 1$ and $S = 0$.
3) Discuss: How can you achieve fairness assumptions (statistical parity, equalized odds) with your classifier? Implement one intervention that should make the classifier more fair and evaluate its effect.

```python
### BUILD CLASSIFIER ###

AGE_COL  = 0
COW_COL  = 1
SCHL_COL = 2
WKHP_COL = 3
SEX_COL  = 4   # 1=male, 2=female

def make_score(X):
    age  = X[:, AGE_COL]  / 100.0
    wkhp = X[:, WKHP_COL] / 100.0
    schl = X[:, SCHL_COL] / 25.0
    cow  = X[:, COW_COL]  / 10.0
    
    # male=1, female=0
    sex_bonus = np.where(X[:, SEX_COL] == 1, 0.2, 0.0)
    
    # equal weight for all 5 features
    score = age + wkhp + schl + cow + sex_bonus
    
    return score

# use a threshold on the score to predict high income
def predict(score, tau):
    return (score >= tau).astype(int)

# accuracy
def accuracy(y_true, y_pred):
    return (y_true == y_pred).mean()

# pick best threshold based on training set accuracy
train_score = make_score(X_train)

taus = np.unique(train_score)  # candidate thresholds
best_tau = None
best_acc = -1.0

for tau in taus:
    y_pred_train = predict(train_score, tau)
    acc = accuracy(y_train, y_pred_train)
    if acc > best_acc:
        best_acc = acc
        best_tau = tau

print("Threshold for best accuracy:", best_tau)
print("Train accuracy at threshold:", best_acc)
```

```python
### EVALUATE ACCURACY ###

# evaluate accuracy on test set
test_score = make_score(X_test)
y_pred_test = predict(test_score, best_tau)
test_acc = accuracy(y_test, y_pred_test)

print("Test accuracy at threshold:", test_acc)
```

```python
### CHECK STATISTICAL PARITY ###
## P(S=1 | G=1) vs P(S=1 | G=2)

S = y_pred_test
T = y_test
G = group_test

# Selection rates per group
p_s_g1 = S[G == 1].mean() # men
p_s_g2 = S[G == 2].mean() # women

print("P(S=1 | G=1) (men):", p_s_g1)
print("P(S=1 | G=2) (women):", p_s_g2)

if np.isclose(p_s_g1, p_s_g2):
    print("Statistical parity satisfied")
else:
    print("Statistical parity NOT satisfied")

```

```python
### CHECK EQUALIZING ODDS ###
# P(S=1 | G=1, T=1) vs P(S=1 | G=2, T=1)  (TPR)
# P(S=1 | G=1, T=0) vs P(S=1 | G=2, T=0)  (FPR)

# TPR 
tpr_g1 = S[(G == 1) & (T == 1)].mean()
tpr_g2 = S[(G == 2) & (T == 1)].mean()

print("TPR (men):   ", tpr_g1)
print("TPR (women): ", tpr_g2)

# FPR 
fpr_g1 = S[(G == 1) & (T == 0)].mean()
fpr_g2 = S[(G == 2) & (T == 0)].mean()

print("FPR (men):   ", fpr_g1)
print("FPR (women): ", fpr_g2)

if np.isclose(tpr_g1, tpr_g2) and np.isclose(fpr_g1, fpr_g2):
    print("Equalizing odds satisfied")
else:
    print("Equalizing odds NOT satisfied")
```

```python
### CHECK EQUALIZING OUTCOMES ###
# P(T=1 | G=1, S=1) vs P(T=1 | G=2, S=1)
# P(T=1 | G=1, S=0) vs P(T=1 | G=2, S=0)

# P(T=1 | G=1, S=1)
p_t1_s1_g1 = T[(G == 1) & (S == 1)].mean()
p_t1_s1_g2 = T[(G == 2) & (S == 1)].mean()

print("P(T=1 | G=1, S=1) (men):", p_t1_s1_g1)
print("P(T=1 | G=2, S=1) (women):", p_t1_s1_g2)

# P(T=1 | G=1, S=0)
p_t1_s0_g1 = T[(G == 1) & (S == 0)].mean()
p_t1_s0_g2 = T[(G == 2) & (S == 0)].mean()

print("P(T=1 | G=1, S=0) (men):", p_t1_s0_g1)
print("P(T=1 | G=2, S=0) (women):", p_t1_s0_g2)

if np.isclose(p_t1_s1_g1, p_t1_s1_g2) and np.isclose(p_t1_s0_g1, p_t1_s0_g2):
    print("Equalizing outcomes satisfied")
else:
    print("Equalizing outcomes NOT satisfied")
```

### Attempt 1 at more fairness: Remove SEX as feature in classifier.
One attempt to make the classifier more fair is to only use AGE and EDUCATION to compute the score. This makes the model group-blind, meaning that it does not explicitly use the protected attribute in decision-making. However, other features such as education level and age may correlate with sex, and disparities may still remain.

```python
def make_score_no_sex(X):
    age  = X[:, AGE_COL]  / 100.0
    wkhp = X[:, WKHP_COL] / 100.0
    schl = X[:, SCHL_COL] / 25.0
    cow  = X[:, COW_COL]  / 10.0
    
    # equal weight for all 5 features
    score = age + wkhp + schl + cow
    
    return score

# find new best threshold
train_score_no_sex = make_score_no_sex(X_train)

taus_no_sex = np.unique(train_score_no_sex)  # candidate thresholds
best_tau_no_sex = None
best_acc_no_sex = -1.0

for tau in taus_no_sex:
    y_pred_train = predict(train_score_no_sex, tau)
    acc = accuracy(y_train, y_pred_train)
    if acc > best_acc_no_sex:
        best_acc_no_sex = acc
        best_tau_no_sex = tau


print("Best threshold (no SEX):", best_tau)
print("Train accuracy:", best_acc)

# evaluate accuracy on test set
test_score = make_score_no_sex(X_test)
y_pred_test = predict(test_score, best_tau)

print("Test accuracy:", accuracy(y_test, y_pred_test))

# check statistical parity again
S = y_pred_test
G = group_test

p_s_g1 = S[G == 1].mean() # men
p_s_g2 = S[G == 2].mean() # women
print("P(S=1 | G=1) (men):", p_s_g1)
print("P(S=1 | G=2) (women):", p_s_g2)

if np.isclose(p_s_g1, p_s_g2):
    print("Statistical parity satisfied")
else:    
    print("Statistical parity NOT satisfied")
    
# check equalizing odds again
T = y_test
tpr_g1 = S[(G == 1) & (T == 1)].mean()
tpr_g2 = S[(G == 2) & (T == 1)].mean()
fpr_g1 = S[(G == 1) & (T == 0)].mean()
fpr_g2 = S[(G == 2) & (T == 0)].mean()

print("TPR (men):   ", tpr_g1)
print("TPR (women): ", tpr_g2)
print("FPR (men):   ", fpr_g1)
print("FPR (women): ", fpr_g2)  

if np.isclose(tpr_g1, tpr_g2) and np.isclose(fpr_g1, fpr_g2):
    print("Equalizing odds satisfied")
else:
    print("Equalizing odds NOT satisfied")

# check equalizing outcomes again
p_t1_s1_g1 = T[(G == 1) & (S == 1)].mean()
p_t1_s1_g2 = T[(G == 2) & (S == 1)].mean()
p_t1_s0_g1 = T[(G == 1) & (S == 0)].mean()
p_t1_s0_g2 = T[(G == 2) & (S == 0)].mean()      

print("P(T=1 | G=1, S=1) (men):", p_t1_s1_g1)
print("P(T=1 | G=2, S=1) (women):", p_t1_s1_g2)
print("P(T=1 | G=1, S=0) (men):", p_t1_s0_g1)
print("P(T=1 | G=2, S=0) (women):", p_t1_s0_g2) 

if np.isclose(p_t1_s1_g1, p_t1_s1_g2) and np.isclose(p_t1_s0_g1, p_t1_s0_g2):
    print("Equalizing outcomes satisfied")
else:
    print("Equalizing outcomes NOT satisfied")

```

After adjusting the normalization and reducing the influence of SEX, the classifier became more balanced. Test accuracy remained high (≈ 0.77), while disparities between men and women decreased substantially compared to earlier versions.

Statistical parity differences are now small (about 2 percentage points). TPR and FPR are also relatively close across groups, although equalized odds is still not fully satisfied. Precision differs somewhat between groups, reflecting differences in underlying income distributions.

This demonstrates that proper normalization and balanced feature scaling can significantly reduce extreme disparities. However, even with a well-balanced score, fairness criteria are not automatically satisfied due to differences in base rates and structural patterns in the data.

### Attempt 2 at more fairness: Grouo-specific thresholds
Same score function
Different threshold for men and women

```python
train_score = make_score(X_train)

score_men   = train_score[group_train == 1]
score_women = train_score[group_train == 2]

baseline_tau = best_tau  # previously found global best threshold
r = predict(train_score, baseline_tau).mean()

print("Target selection rate r (from baseline):", r)

# Find group-specific thresholds that achieve selection rate r on TRAIN
# Selecting S=1 means score >= tau, so tau should be the (1-r)-quantile.
tau_men   = np.quantile(score_men,   1 - r)
tau_women = np.quantile(score_women, 1 - r)

print("tau_men:", tau_men)
print("tau_women:", tau_women)

# predict on TRAIN with group-specific thresholds
def predict_group_threshold(score, group, tau_men, tau_women):
    y_pred = np.zeros_like(score, dtype=int)
    y_pred[group == 1] = predict(score[group == 1], tau_men)
    y_pred[group == 2] = predict(score[group == 2], tau_women)
    return y_pred

# --- train predictions ---
y_pred_train_dp = predict_group_threshold(train_score, group_train, tau_men, tau_women)

print("\nTRAIN accuracy (DP thresholds):", accuracy(y_train, y_pred_train_dp))
print("P(S=1|men) train:", y_pred_train_dp[group_train == 1].mean())
print("P(S=1|women) train:", y_pred_train_dp[group_train == 2].mean())

# --- test predictions ---
test_score = make_score(X_test)
y_pred_test_dp = predict_group_threshold(test_score, group_test, tau_men, tau_women)

print("\nTEST accuracy (DP thresholds):", accuracy(y_test, y_pred_test_dp))
print("P(S=1|men) test:", y_pred_test_dp[group_test == 1].mean())
print("P(S=1|women) test:", y_pred_test_dp[group_test == 2].mean())


# check how it affects equalizing odds and outcomes
# --- Equalized Odds (on TEST) ---

def rate(mask):
    return y_pred_test_dp[mask].mean()

print("TPR (men):   ", rate((group_test==1) & (y_test==1)))
print("TPR (women): ", rate((group_test==2) & (y_test==1)))

print("FPR (men):   ", rate((group_test==1) & (y_test==0)))
print("FPR (women): ", rate((group_test==2) & (y_test==0)))

print("P(T=1 | S=1, men):",
      y_test[(group_test==1) & (y_pred_test_dp==1)].mean())

print("P(T=1 | S=1, women):",
      y_test[(group_test==2) & (y_pred_test_dp==1)].mean())

print("P(T=1 | S=0, men):",
      y_test[(group_test==1) & (y_pred_test_dp==0)].mean())

print("P(T=1 | S=0, women):",
      y_test[(group_test==2) & (y_pred_test_dp==0)].mean())
```

### Discussion of results

To reduce demographic disparity, we introduced group-specific thresholds. Since the original model tended to favor men (men generally had slightly higher scores), we set a higher threshold for men and a lower threshold for women. This adjustment ensured that both groups achieved approximately the same selection rate.

As intended, this intervention significantly improved statistical parity, because the thresholds were explicitly chosen to align selection rates across groups. The selection gap became very small on both the training and test data.

However, this intervention did not improve the other fairness metrics. Equalized odds (TPR and FPR) still differ between groups, and predictive parity (equalized outcome) remains violated. This demonstrates an important principle in algorithmic fairness:

Enforcing one fairness criterion does not automatically satisfy others.

In fact, different fairness definitions can conflict with each other, and it is generally impossible to satisfy all of them simultaneously when base rates differ across groups. This experiment clearly illustrates the fairness trade-off: we can improve demographic parity, but we cannot “have it all.”
